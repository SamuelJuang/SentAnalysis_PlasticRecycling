```python
import re

def remove_links(text):
    return re.sub(r'http\S+|www\S+|https\S+', '', text)
```

```python
import requests
import time

def FacebookAPICall(search, limit):
    url = "https://facebook-scraper3.p.rapidapi.com/search/posts"
    headers = {
        "x-rapidapi-key": "564af8b867mshb69c83d38872dfep10ddddjsn3981c2299832",
        "x-rapidapi-host": "facebook-scraper3.p.rapidapi.com"
    }
    cursor = None 
    all_posts = []
    while len(all_posts) < limit:
        params = {"query": search}
        if cursor:
            params["cursor"] = cursor
        response = requests.get(url, headers=headers, params=params)
        if response.status_code != 200:
            print("Error:", response.status_code, response.text)
            break
        data = response.json()
        posts = data.get("results", [])
        all_posts.extend(posts)
        print(f"Fetched {len(posts)} posts (Total: {len(all_posts)})")
        cursor = data.get("cursor")
        if not cursor:
            print("No more pages.")
            break
        time.sleep(1)
    print(f"Total collected posts: {len(all_posts)}")
    return all_posts
```
