```python
import pandas as pd

bisnis_df = pd.read_csv("analysis/edukasi_anak_60_sampled.csv")
bisnis_df = bisnis_df[:55]
```

```python
bisnis_df = bisnis_df.rename(columns={
    'sentimenteffective': 'Recyclability',
    'sentimentpet': 'Recyclability (PET)',
    'sentimentprocess': 'Recycling',
    'sentimentrecycle': 'Future'
})
```

```python
labels = ['Recyclability', 'Recyclability (PET)', 'Recycling', 'Future']

bisnis_df = pd.get_dummies(bisnis_df, columns=labels, dtype=int)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# bisnis_df.rename(columns={'sentimenteffective': 'Recyclability'}, inplace=True)
# bisnis_df.rename(columns={'sentimentpet': 'Recyclability (PET)'}, inplace=True)
# bisnis_df.rename(columns={'sentimentprocess': 'Recycling'}, inplace=True)
# bisnis_df.rename(columns={'sentimentrecycle': 'Future'}, inplace=True)
# df_encoded = pd.get_dummies(bisnis_df, columns=['Recyclability', 'Recyclability (PET)', 'Recycling', 'Future'])
label_sums = bisnis_df[[
        "Recyclability_Negative", "Recyclability_Neutral", "Recyclability_Positive",
        "Recyclability (PET)_Negative", "Recyclability (PET)_Neutral", "Recyclability (PET)_Positive",
        "Recycling_Negative", "Recycling_Neutral", "Recycling_Positive",
        "Future_Negative", "Future_Neutral", "Future_Positive"
    ]].sum()

label_df = label_sums.reset_index()
label_df.columns = ['label', 'count']
label_df['aspect'] = label_df['label'].apply(lambda x: x.split('_')[0])
label_df['sentiment'] = label_df['label'].apply(lambda x: x.split('_')[1])

fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(data=label_df, x='aspect', y='count', hue='sentiment', palette='Set2', ax=ax)
ax.set_title("Data Count by label and Sentiment")
ax.set_ylabel("Number of Samples")
ax.set_xlabel("Aspect")
ax.legend(title='Sentiment')
```
